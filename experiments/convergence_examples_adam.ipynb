# Метод Adam

```python
import sys
from pathlib import Path

src_path = Path("../src").resolve()
sys.path.append(str(src_path))

import numpy as np
from typing import Callable, List
import math
from adam import adam
from mytypes import Array
from show import show
```

```python
def unbound_func_adam(x: Array) -> float:
    return float(np.sum(x**2))


def grad_unbound_func_adam(x: Array) -> Array:
    return 2.0 * x


alpha: Callable[[int], float] = lambda k: 0.1
beta1: Callable[[int], float] = lambda k: 0.9
beta2: Callable[[int], float] = lambda k: 0.999
epsilon: float = 1e-8

x0: Array = np.array([10.0], dtype=np.float64)
T: int = 500

x_vals: List[Array] = adam(
    grad_unbound_func_adam,
    x0,
    T,
    alpha=alpha,
    beta1=beta1,
    beta2=beta2,
    epsilon=epsilon,
)
f_vals: List[float] = [unbound_func_adam(x) for x in x_vals]

show(x_vals, f_vals, ylabel2=r"$f(x) = x^2$")
```

```python
def nonconv_func_adam(x: Array) -> float:
    return x[0] ** 2 + math.sin(3.0 * x[0])


def grad_nonconv_func_adam(x: Array) -> Array:
    return np.array(
        [2.0 * x[0] + 3.0 * math.cos(3.0 * x[0])], dtype=np.float64
    )


x0: Array = np.array([5.0], dtype=np.float64)
T: int = 150

alpha: Callable[[int], float] = lambda k: 0.2
beta1: Callable[[int], float] = lambda k: 0.9
beta2: Callable[[int], float] = lambda k: 0.999
epsilon: float = 1e-8

x_vals: List[Array] = adam(
    grad_nonconv_func_adam,
    x0,
    T,
    alpha=alpha,
    beta1=beta1,
    beta2=beta2,
    epsilon=epsilon,
)
f_vals: List[float] = [nonconv_func_adam(x) for x in x_vals]

show(
    x_vals,
    f_vals,
    ylabel2=r"$f(x) = x^2 + \sin(3x)$",
    x_min=-0.427,
    f_min=-0.776,
)
```
